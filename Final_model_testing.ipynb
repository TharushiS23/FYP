```python
import os

test_root = "/content/FYP/Split/test"  
class_names = ["512Glioma", "512Meningioma", "512Normal", "512Pituitary"]

# Make a list of all test images
test_images = []
for cls in class_names:
    cls_path = os.path.join(test_root, cls)
    for img_name in os.listdir(cls_path):
        test_images.append({
            "path": os.path.join(cls_path, img_name),
            "true_class": cls
        })

# check first 5 entries
test_images[:5]
```

```python
all_preds = []
all_labels = []

for img_info in test_images:
    pred_class = Ensemble_model.predict(img_info["path"])
    all_preds.append(pred_class)
    all_labels.append(img_info["true_class"])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(all_labels, all_preds)
print(f"Test set accuracy: {accuracy*100:.2f}%")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Overall accuracy
test_acc = accuracy_score(all_labels, all_preds)
print(f"Test Accuracy: {test_acc * 100:.2f}%\n")

# Detailed per-class metrics
print("Per-class metrics for the ensemble model:")
print(classification_report(all_labels, all_preds, target_names=class_names))
```

```python
from sklearn.metrics import confusion_matrix

class_to_idx = {name: i for i, name in enumerate(class_names)}
y_true = [class_to_idx[c] for c in all_labels]
y_pred = [class_to_idx[c] for c in all_preds]

cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:")
print(cm)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=class_names, yticklabels=class_names, cmap="RdPu")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()
```

```python
import sys
sys.path.append("/content/FYP")  # folder where Tuned_model_ensemble.py 
import Tuned_model_ensemble
```

```python
import os

test_root_tuned_ensemble = "/content/FYP/Split/test"
class_names_tuned_ensemble = ["512Glioma", "512Meningioma", "512Normal", "512Pituitary"]

test_images_tuned_ensemble = []
for class_name_tuned_ensemble in class_names_tuned_ensemble:
    class_path_tuned_ensemble = os.path.join(test_root_tuned_ensemble, class_name_tuned_ensemble)
    for image_name_tuned_ensemble in os.listdir(class_path_tuned_ensemble):
        test_images_tuned_ensemble.append({
            "path": os.path.join(class_path_tuned_ensemble, image_name_tuned_ensemble),
            "true_class": class_name_tuned_ensemble
        })

print("Total test images:", len(test_images_tuned_ensemble))
print(test_images_tuned_ensemble[:5])
```

```python
from Tuned_model_ensemble import predict_tuned_ensemble
```

```python
all_preds_tuned_ensemble = []
all_labels_tuned_ensemble = []

for img_info_tuned_ensemble in test_images_tuned_ensemble:
    predicted_class_tuned_ensemble = predict_tuned_ensemble(img_info_tuned_ensemble["path"])
    all_preds_tuned_ensemble.append(predicted_class_tuned_ensemble)
    all_labels_tuned_ensemble.append(img_info_tuned_ensemble["true_class"])
```

```python
from sklearn.metrics import accuracy_score

accuracy_tuned_ensemble = accuracy_score(all_labels_tuned_ensemble, all_preds_tuned_ensemble)
print(f"Tuned ensemble test accuracy: {accuracy_tuned_ensemble * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report

test_accuracy_tuned_ensemble = accuracy_score(all_labels_tuned_ensemble, all_preds_tuned_ensemble)
print(f"Test Accuracy: {test_accuracy_tuned_ensemble * 100:.2f}%\n")

print("Per-class metrics for the tuned ensemble model:")
print(
    classification_report(
        all_labels_tuned_ensemble,
        all_preds_tuned_ensemble,
        target_names=class_names_tuned_ensemble
    )
)
```

```python
from sklearn.metrics import confusion_matrix

class_to_idx_tuned_ensemble = {
    class_name_tuned_ensemble: idx_tuned_ensemble
    for idx_tuned_ensemble, class_name_tuned_ensemble in enumerate(class_names_tuned_ensemble)
}

y_true_tuned_ensemble = [
    class_to_idx_tuned_ensemble[class_label_tuned_ensemble]
    for class_label_tuned_ensemble in all_labels_tuned_ensemble
]

y_pred_tuned_ensemble = [
    class_to_idx_tuned_ensemble[class_pred_tuned_ensemble]
    for class_pred_tuned_ensemble in all_preds_tuned_ensemble
]

confusion_matrix_tuned_ensemble = confusion_matrix(
    y_true_tuned_ensemble,
    y_pred_tuned_ensemble
)

print("Confusion Matrix for tuned ensemble:")
print(confusion_matrix_tuned_ensemble)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(
    confusion_matrix_tuned_ensemble,
    annot=True,
    fmt='d',
    xticklabels=class_names_tuned_ensemble,
    yticklabels=class_names_tuned_ensemble,
    cmap="RdPu"
)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix - Tuned Ensemble Model")
plt.show()
```
